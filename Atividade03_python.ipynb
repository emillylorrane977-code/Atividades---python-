```python


 1. Guardando e analisando notas

python

 Ler 5 notas e armazenar em lista
notas = []
for i in range(5):
    nota = float(input("Digite uma nota: "))
    notas.append(nota)

 Calcular média
soma = 0
for nota in notas:
    soma += nota
media = soma / len(notas)

 Contar notas >= média
acima_ou_igual = 0
for nota in notas:
    if nota >= media:
        acima_ou_igual += 1

 Exibir resultados
print("Notas:", notas)
print("Média:", media)
print("Notas >= média:", acima_ou_igual)





 2. Coordenada representada por tupla

python

 Leitura dos valores
x = int(input("x: "))
y = int(input("y: "))

 Armazena em TUPLA (imutável)
ponto = (x, y)

 Classificação do ponto
if ponto[0] == 0 and ponto[1] == 0:
    local = "Origem"
elif ponto[0] == 0:
    local = "Eixo Y"
elif ponto[1] == 0:
    local = "Eixo X"
elif ponto[0] > 0 and ponto[1] > 0:
    local = "1º quadrante"
elif ponto[0] < 0 and ponto[1] > 0:
    local = "2º quadrante"
elif ponto[0] < 0 and ponto[1] < 0:
    local = "3º quadrante"
else:
    local = "4º quadrante"

print("Ponto:", ponto)
print("Local:", local)




 3. Cadastro simples de produto

python

 Dicionário vazio
produto = {}

 Ler dados
produto["nome"] = input("Nome: ")
produto["preco"] = float(input("Preço: "))
produto["quantidade"] = int(input("Quantidade: "))

 Calcular valor total
total = produto["preco"] * produto["quantidade"]

 Exibir
print("Produto:", produto)
print("Valor em estoque:", total)






 4. Lista de números: pares e ímpares

python

 Listas
numeros = []
pares = []
impares = []

 Ler 10 números
for i in range(10):
    n = int(input(f"Digite o {i+1}º número: "))
    numeros.append(n)

 Separar pares e ímpares
for num in numeros:
    if num % 2 == 0:
        pares.append(num)
    else:
        impares.append(num)

 Exibir
print("\nTodos os números:", numeros)
print("Pares:", pares)
print("Ímpares:", impares)




 5. Maior e menor da lista

python

 Ler 8 números
numeros = []
for i in range(8):
    num = int(input(f"Digite o {i+1}º número: "))
    numeros.append(num)

Inicializar com o primeiro valor
maior = numeros[0]
menor = numeros[0]
pos_maior = 0
pos_menor = 0

 Percorrer comparando — SEM usar max() e min()
for indice in range(1, len(numeros)):
    if numeros[indice] > maior:
        maior = numeros[indice]
        pos_maior = indice
    if numeros[indice] < menor:
        menor = numeros[indice]
        pos_menor = indice

 Exibir
print("\nLista:", numeros)
print(f"Maior valor: {maior} — Posição: {pos_maior}")
print(f"Menor valor: {menor} — Posição: {pos_menor}")




 6. Nomes sem repetição

python

nomes = []

 Ler nomes até digitar FIM
while True:
    entrada = input("Digite um nome (FIM para encerrar): ")
    if entrada.upper() == "FIM":
        break
     Adicionar SOMENTE se não estiver na lista
    if entrada not in nomes:
        nomes.append(entrada)
    else:
        print("Esse nome já está cadastrado!")

 Exibir
print("\nNomes cadastrados:", nomes)
print("Quantidade de nomes diferentes:", len(nomes))




 7. Dias da semana em uma tupla

python

 Tupla — dados fixos
dias_semana = ("Segunda-feira", "Terça-feira", "Quarta-feira",
               "Quinta-feira", "Sexta-feira", "Sábado", "Domingo")

 Ler número
num = int(input("Digite um número de 1 a 7: "))

Validar e exibir
if 1 <= num <= 7:
    print(f"Dia correspondente: {dias_semana[num - 1]}")
else:
    print("Opção inválida!")




 8. Estatísticas de uma tupla

python

Ler 6 números e formar tupla
valores = []
for i in range(6):
    n = int(input(f"Digite o {i+1}º número: "))
    valores.append(n)
numeros = tuple(valores)  # converte para tupla

Contar número 5, posição da 1ª ocorrência e contar pares
cont_cinco = 0
pos_cinco = -1
cont_pares = 0

for indice, num in enumerate(numeros):
    if num == 5:
        cont_cinco += 1
        if pos_cinco == -1:
            pos_cinco = indice
    if num % 2 == 0:
        cont_pares += 1

 Exibir
print("\nTupla:", numeros)
print(f"Quantas vezes o número 5 apareceu: {cont_cinco}")
if pos_cinco != -1:
    print(f"Posição da primeira ocorrência do 5: {pos_cinco}")
else:
    print("O número 5 não está na tupla.")
print(f"Quantidade de valores pares: {cont_pares}")




 9. Dicionário de aluno

python

 Cadastro
aluno = {}
aluno["nome"] = input("Nome do aluno: ")
aluno["turma"] = input("Turma: ")
aluno["nota1"] = float(input("Nota 1: "))
aluno["nota2"] = float(input("Nota 2: "))

 Calcular média e situação
aluno["media"] = (aluno["nota1"] + aluno["nota2"]) / 2

if aluno["media"] >= 7:
    aluno["situacao"] = "Aprovado"
elif aluno["media"] >= 5:
    aluno["situacao"] = "Recuperação"
else:
    aluno["situacao"] = "Reprovado"

 Exibir tudo
print("\nDados do aluno:")
for chave, valor in aluno.items():
    print(f"{chave.capitalize()}: {valor}")




 10. Consulta de preços

python

 Dicionário de produtos
precos = {
    "caderno": 15.50,
    "caneta": 2.50,
    "lápis": 1.20,
    "borracha": 3.00,
    "mochila": 89.90
}

 Consulta
produto_busca = input("Digite o nome do produto: ").lower()

 Direto com chave — sem vários if
if produto_busca in precos:
    print(f"Preço de {produto_busca}: R$ {precos[produto_busca]:.2f}")
else:
    print(f"Produto '{produto_busca}' não cadastrado.")




11. Frequência das palavras

python

contagem = {}

Ler 8 palavras
for i in range(8):
    palavra = input(f"Digite a {i+1}ª palavra: ").lower()
    if palavra in contagem:
        contagem[palavra] += 1
    else:
        contagem[palavra] = 1

 Exibir resultado
print("\nFrequência das palavras:")
for palavra, qtd in contagem.items():
    print(f"'{palavra}': {qtd} vez(es)")




12. Estoque e atualização

python

# Estoque inicial
estoque = {
    "caderno": 50,
    "caneta": 120,
    "lápis": 75,
    "borracha": 30
}

Entrada
produto = input("Nome do produto vendido: ").lower()
quantidade = int(input("Quantidade vendida: "))

Verificar e processar
if produto in estoque:
    if estoque[produto] >= quantidade:
        estoque[produto] -= quantidade
        print(f" Venda realizada! Estoque atualizado: {estoque[produto]} unidades.")
    else:
        print(f" Estoque insuficiente! Disponível: {estoque[produto]} unidades.")
else:
    print(f" Produto '{produto}' não cadastrado.")

 Mostrar estoque final
print("\n Estoque final:")
for nome, qtd in estoque.items():
    print(f"{nome}: {qtd}")




 13. Cadastro de vários alunos

python

turma = []  # Lista com todos os alunos

 Cadastrar 5 alunos
for i in range(5):
    print(f"\nAluno {i+1}:")
    aluno = {}  # Dicionário para cada aluno
    aluno["nome"] = input("Nome: ")
    aluno["nota1"] = float(input("Nota 1: "))
    aluno["nota2"] = float(input("Nota 2: "))
    aluno["media"] = (aluno["nota1"] + aluno["nota2"]) / 2
    turma.append(aluno)

 Exibir todos
print("\n" + "="*40)
print("Boletim da Turma")
print("="*40)
for aluno in turma:
    print(f"Nome: {aluno['nome']} — Média: {aluno['media']:.1f}")




 14. Catálogo com código imutável

python

catalogo = []  # Lista de tuplas

Cadastrar 5 produtos
for i in range(5):
    print(f"\nProduto {i+1}:")
    codigo = input("Código: ")
    nome = input("Nome: ")
    preco = float(input("Preço: R$ "))
    produto = (codigo, nome, preco)  # TUPLA = dados imutáveis
    catalogo.append(produto)

 Busca
print("\n" + "-"*40)
busca = input("Digite o código para buscar: ")

encontrado = False
for prod in catalogo:
    cod, nom, prc = prod
    if cod == busca:
        print(f"\n Encontrado!")
        print(f"Nome: {nom}")
        print(f"Preço: R$ {prc:.2f}")
        encontrado = True
        break

if not encontrado:
    print("\n Código não encontrado.")




 15. Mini sistema de estoque

python

estoque = []  # Lista de dicionários

 Cadastrar 5 itens
for i in range(5):
    print(f"\nItem {i+1} de 5:")
    item = {}
    item["codigo"] = input("Código: ")
    item["nome"] = input("Nome: ")
    item["quantidade"] = int(input("Quantidade: "))
    item["preco"] = float(input("Preço unitário: R$ "))
    estoque.append(item)

 (a) Mostrar todos os itens
print("\n" + "="*60)
print(" Itens Cadastrados")
print("="*60)
for prod in estoque:
    print(f"Código: {prod['codigo']} | Nome: {prod['nome']} | Qtd: {prod['quantidade']} | Preço: R$ {prod['preco']:.2f}")

 (b) Valor total do estoque
valor_total = 0
qtd_total = 0
for prod in estoque:
    valor_total += prod["quantidade"] * prod["preco"]
    qtd_total += prod["quantidade"]
print("\n Valor total do estoque: R$", f"{valor_total:.2f}")

(c) Item de maior valor total — SEM usar max()
maior_item = estoque[0]
maior_valor = estoque[0]["quantidade"] * estoque[0]["preco"]
for prod in estoque[1:]:
    total = prod["quantidade"] * prod["preco"]
    if total > maior_valor:
        maior_valor = total
        maior_item = prod
print(f" Produto de maior valor: {maior_item['nome']} — R$ {maior_valor:.2f}")

(d) Tupla resumo
resumo = (qtd_total, valor_total)
print("\n Resumo da tupla:")
print(f"Total de itens: {resumo[0]}")
print(f"Valor total: R$ {resumo[1]:.2f}")
```

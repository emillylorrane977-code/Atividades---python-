```python



Exercício 1 — Positivo, Negativo ou Zero

python

numero = int(input("Digite um número: "))
if numero > 0:
    print("Positivo")
elif numero < 0:
    print("Negativo")
else:
    print("Zero")




Exercício 2 — Maior entre Dois Números

python

a = int(input("Digite o primeiro número: "))
b = int(input("Digite o segundo número: "))
if a > b:
    print("Maior:", a)
elif b > a:
    print("Maior:", b)
else:
    print("Os números são iguais.")




Exercício 3 — Somando Vários Números

python

soma = 0
for i in range(5):
    numero = int(input("Digite um número: "))
    soma = soma + numero
print("Soma:", soma)




Exercício 4 — Par ou Ímpar

python

numero = int(input("Digite um número inteiro: "))
if numero % 2 == 0:
    print("Par")
else:
    print("Ímpar")




Exercício 5 — Situação do Aluno

python

nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
media = (nota1 + nota2) / 2
print(f"Média: {media:.1f}")
if media >= 7:
    print("Aprovado")
elif media >= 5:
    print("Recuperação")
else:
    print("Reprovado")




Exercício 6 — Maior entre Três Números

python

n1 = int(input("Primeiro número: "))
n2 = int(input("Segundo número: "))
n3 = int(input("Terceiro número: "))
maior = n1
if n2 > maior:
    maior = n2
if n3 > maior:
    maior = n3
print("Maior número:", maior)




Exercício 7 — Contagem Crescente

python

N = int(input("Digite um número inteiro positivo: "))
for numero in range(1, N + 1):
    print(numero)




Exercício 8 — Soma de 1 até N

python

N = int(input("Digite N: "))
soma = 0
for numero in range(1, N + 1):
    soma = soma + numero
print("Soma:", soma)




Exercício 9 — Contar Números Pares

python

contador_pares = 0
for i in range(10):
    num = int(input(f"Digite o número {i+1}: "))
    if num % 2 == 0:
        contador_pares = contador_pares + 1
print(f"Foram digitados {contador_pares} números pares.")




Exercício 10 — Soma Apenas dos Positivos

python

soma_positivos = 0
for i in range(8):
    valor = int(input(f"Digite o valor {i+1}: "))
    if valor > 0:
        soma_positivos = soma_positivos + valor
print("Soma dos positivos:", soma_positivos)




Exercício 11 — Maior Valor Digitado

python

maior = int(input("Digite o 1º número: "))
for i in range(2, 7):
    numero = int(input(f"Digite o {i}º número: "))
    if numero > maior:
        maior = numero
print("Maior valor:", maior)




Exercício 12 — Tabuada

python

numero = int(input("Digite um número para ver a tabuada: "))
for i in range(1, 11):
    print(f"{numero} x {i} = {numero * i}")




Exercício 13 — Senha

python

senha_correta = "1234"
senha_digitada = input("Digite a senha: ")
while senha_digitada != senha_correta:
    print("Senha incorreta! Tente novamente.")
    senha_digitada = input("Digite a senha: ")
print("Acesso permitido!")




Exercício 14 — Número Secreto

python

numero_secreto = 7
tentativas = 0
while True:
    chute = int(input("Tente adivinhar o número secreto: "))
    tentativas = tentativas + 1
    if chute == numero_secreto:
        print("Parabéns! Você acertou.")
        print(f"Foram necessárias {tentativas} tentativas.")
        break
    elif chute < numero_secreto:
        print("Tente um número maior.")
    else:
        print("Tente um número menor.")




Exercício 15 — Sequência de Números

python

qtd_total = 0
qtd_positivos = 0
qtd_negativos = 0
soma_total = 0

numero = int(input("Digite um número: "))
while numero != 0:
    qtd_total = qtd_total + 1
    if numero > 0:
        qtd_positivos = qtd_positivos + 1
    elif numero < 0:
        qtd_negativos = qtd_negativos + 1
    soma_total = soma_total + numero
    numero = int(input("Digite um número: "))

print(f"Quantidade de números digitados: {qtd_total}")
print(f"Quantidade de números positivos: {qtd_positivos}")
print(f"Quantidade de números negativos: {qtd_negativos}")
print(f"Soma de todos os números digitados: {soma_total}
```

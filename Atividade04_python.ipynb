```python


EXERCÍCIO 1 — Analisando Temperaturas (Lista)

print("="*60)
print("EXERCÍCIO 1 — Analisando Temperaturas")
print("="*60)

Armazenar as 7 temperaturas
temperaturas = []
for i in range(7):
    temp = float(input(f"Temperatura {i+1}: "))
    temperaturas.append(temp)

 Calcular soma e média
soma = 0
for t in temperaturas:
    soma += t
media = soma / len(temperaturas)

 Classificar valores
acima = abaixo = iguais = 0
for t in temperaturas:
    if t > media:
        acima += 1
    elif t < media:
        abaixo += 1
    else:
        iguais += 1

 Exibir resultados
print("\n Resultados:")
print(f"Temperaturas: {temperaturas}")
print(f"Média: {media:.2f}")
print(f"Acima da média: {acima}")
print(f"Abaixo da média: {abaixo}")
print(f"Iguais à média: {iguais}")
print("\nExercício 1 concluído!\n")



 EXERCÍCIO 2 — Ponto e Distância da Origem (Tupla)

print("="*60)
print("EXERCÍCIO 2 — Ponto e Distância da Origem")
print("="*60)

x = int(input("x: "))
y = int(input("y: "))
ponto = (x, y)

# Classificação do ponto
if ponto[0] == 0 and ponto[1] == 0:
    local = "Origem"
elif ponto[0] == 0:
    local = "Eixo Y"
elif ponto[1] == 0:
    local = "Eixo X"
elif ponto[0] > 0 and ponto[1] > 0:
    local = "1º quadrante"
elif ponto[0] < 0 and ponto[1] > 0:
    local = "2º quadrante"
elif ponto[0] < 0 and ponto[1] < 0:
    local = "3º quadrante"
else:
    local = "4º quadrante"

Cálculo da distância ao quadrado
dist2 = ponto[0] ** 2 + ponto[1] ** 2

Exibir resultados
print(f"\nPonto: {ponto}")
print(f"Local: {local}")
print(f"Distância²: {dist2}")
print(" Exercício 2 concluído!\n")


EXERCÍCIO 3 — Produto com Controle de Estoque (Dicionário)

print("="*60)
print("EXERCÍCIO 3 — Produto com Controle de Estoque")
print("="*60)

produto = {}
produto["nome"] = input("Nome: ")
produto["preco"] = float(input("Preço: "))
produto["quantidade"] = int(input("Quantidade: "))

 Calcular valor total em estoque
produto["valor_estoque"] = produto["preco"] * produto["quantidade"]

Exibir dicionário completo
print(f"\nProduto: {produto}")
print("Exercício 3 concluído!\n")


EXERCÍCIO 4 — Maior Sequência de Crescimento (Lista)

print("="*60)
print("EXERCÍCIO 4 — Maior Sequência de Crescimento")
print("="*60)

numeros = []
for _ in range(10):
    numeros.append(int(input("Número: ")))

maior_seq = 1
atual = 1

for i in range(1, len(numeros)):
    if numeros[i] > numeros[i-1]:
        atual += 1
        if atual > maior_seq:
            maior_seq = atual
    else:
        atual = 1

print(f"\nNúmeros: {numeros}")
print(f"Maior sequência crescente consecutiva: {maior_seq}")
print("Exercício 4 concluído!\n")


EXERCÍCIO 5 — Pontos no Plano (Lista + Tupla)

print("="*60)
print("EXERCÍCIO 5 — Pontos no Plano")
print("="*60)

pontos = []
qtd_quadrantes = [0, 0, 0, 0]  # 1º, 2º, 3º, 4º
maior_dist = 0
ponto_maior = None

for _ in range(6):
    x = int(input("x: "))
    y = int(input("y: "))
    ponto = (x, y)
    pontos.append(ponto)

    Classificar quadrante
    if x > 0 and y > 0:
        qtd_quadrantes[0] += 1
    elif x < 0 and y > 0:
        qtd_quadrantes[1] += 1
    elif x < 0 and y < 0:
        qtd_quadrantes[2] += 1
    elif x > 0 and y < 0:
        qtd_quadrantes[3] += 1

    Calcular distância ao quadrado
    dist = x**2 + y**2
    if dist > maior_dist:
        maior_dist = dist
        ponto_maior = ponto

print(f"\nPontos: {pontos}")
print(f"1º quadrante: {qtd_quadrantes[0]}")
print(f"2º quadrante: {qtd_quadrantes[1]}")
print(f"3º quadrante: {qtd_quadrantes[2]}")
print(f"4º quadrante: {qtd_quadrantes[3]}")
print(f"Ponto com maior x²+y²: {ponto_maior} ({maior_dist})")
print("Exercício 5 concluído!\n")


 EXERCÍCIO 6 — Frequência e Palavra Mais Repetida (Dicionário)

print("="*60)
print("EXERCÍCIO 6 — Frequência e Palavra Mais Repetida")
print("="*60)

frequencia = {}

for _ in range(10):
    palavra = input("Palavra: ").strip()
    if palavra in frequencia:
        frequencia[palavra] += 1
    else:
        frequencia[palavra] = 1

 Encontrar maior frequência sem usar max()
maior_qtd = 0
for p in frequencia:
    if frequencia[p] > maior_qtd:
        maior_qtd = frequencia[p]

 Palavras com maior frequência
mais_frequentes = []
for p in frequencia:
    if frequencia[p] == maior_qtd:
        mais_frequentes.append(p)

print("\n Frequência de palavras:")
for p, qtd in frequencia.items():
    print(f"- {p}: {qtd}x")
print(f"\nPalavra(s) mais repetida(s) ({maior_qtd}x): {', '.join(mais_frequentes)}")
print(" Exercício 6 concluído!\n")



EXERCÍCIO 7 — Boletim de uma Turma (Lista + Dicionário)

print("="*60)
print("EXERCÍCIO 7 — Boletim de uma Turma")
print("="*60)

turma = []

 Cadastrar 6 alunos
for _ in range(6):
    nome = input("Nome do aluno: ")
    n1 = float(input("Nota 1: "))
    n2 = float(input("Nota 2: "))
    n3 = float(input("Nota 3: "))
    media = (n1 + n2 + n3) / 3
    turma.append({
        "nome": nome,
        "notas": [n1, n2, n3],
        "media": media
    })

Calcular média geral da turma
soma_geral = 0
for aluno in turma:
    soma_geral += aluno["media"]
media_geral = soma_geral / len(turma)

 Encontrar maior média sem usar max()
maior_media = turma[0]["media"]
for a in turma:
    if a["media"] > maior_media:
        maior_media = a["media"]

alunos_maior_media = []
for a in turma:
    if a["media"] == maior_media:
        alunos_maior_media.append(a["nome"])

alunos_acima_media = []
for a in turma:
    if a["media"] >= media_geral:
        alunos_acima_media.append(a["nome"])

print("\n Boletim da Turma:")
for a in turma:
    print(f"{a['nome']} — Média: {a['media']:.2f}")
print(f"\nMédia geral da turma: {media_geral:.2f}")
print(f"Aluno(s) com maior média: {', '.join(alunos_maior_media)} ({maior_media:.2f})")
print(f"Aluno(s) com média ≥ média da turma: {', '.join(alunos_acima_media)}")
print(" Exercício 7 concluído!\n")



 EXERCÍCIO 8 — Controle de Estoque com Movimentações

print("="*60)
print("EXERCÍCIO 8 — Controle de Estoque com Movimentações")
print("="*60)

estoque = []

Cadastrar 5 produtos
for _ in range(5):
    codigo = input("Código do produto: ")
    nome = input("Nome: ")
    preco = float(input("Preço: "))
    qtd = int(input("Quantidade em estoque: "))
    estoque.append({
        "codigo": codigo,
        "nome": nome,
        "preco": preco,
        "quantidade": qtd
    })

 5 operações de entrada/saída
for op in range(5):
    print(f"\n--- Operação {op+1} ---")
    cod = input("Código do produto: ")
    tipo = input("Entrada (E) ou Saída (S): ").upper()
    qtd_op = int(input("Quantidade: "))

     Localizar produto
    prod_encontrado = None
    for p in estoque:
        if p["codigo"] == cod:
            prod_encontrado = p
            break

    if not prod_encontrado:
        print(" Produto não encontrado! Operação cancelada.")
        continue

    if tipo == "E":
        prod_encontrado["quantidade"] += qtd_op
        print(f" Entrada registrada. Estoque atual: {prod_encontrado['quantidade']}")
    elif tipo == "S":
        if prod_encontrado["quantidade"] >= qtd_op:
            prod_encontrado["quantidade"] -= qtd_op
            print(f" Saída registrada. Estoque atual: {prod_encontrado['quantidade']}")
        else:
            print(" Estoque insuficiente! Saída cancelada.")
    else:
        print(" Tipo inválido! Use E ou S.")

 Relatório final
print("\n Estoque Final:")
valor_total = 0
for p in estoque:
    valor = p["preco"] * p["quantidade"]
    valor_total += valor
    print(f"{p['codigo']} — {p['nome']}: {p['quantidade']} unidades | R$ {valor:.2f}")
print(f" Valor total do estoque: R$ {valor_total:.2f}")
print(" Exercício 8 concluído!\n")



 EXERCÍCIO 9 — Ranking de Jogadores

print("="*60)
print("EXERCÍCIO 9 — Ranking de Jogadores")
print("="*60)

jogadores = []

for _ in range(5):
    nome = input("Nome do jogador: ")
    p1 = int(input("Pontuação 1: "))
    p2 = int(input("Pontuação 2: "))
    p3 = int(input("Pontuação 3: "))
    total = p1 + p2 + p3
    media = total / 3

     Encontrar maior e menor sem max/min
    maior_p = p1
    menor_p = p1
    for p in [p2, p3]:
        if p > maior_p: maior_p = p
        if p < menor_p: menor_p = p
    diferenca = maior_p - menor_p

    jogadores.append({
        "nome": nome, "pontos": [p1, p2, p3],
        "total": total, "media": media, "diferenca": diferenca
    })

 Maior total
maior_total = jogadores[0]["total"]
for j in jogadores:
    if j["total"] > maior_total:
        maior_total = j["total"]

 Menor diferença
menor_diff = jogadores[0]["diferenca"]
for j in jogadores:
    if j["diferenca"] < menor_diff:
        menor_diff = j["diferenca"]

print("\n Ranking:")
for j in jogadores:
    print(f"{j['nome']} — Total: {j['total']} | Média: {j['media']:.1f} | Diferença: {j['diferenca']}")

print("\n Maior total:")
for j in jogadores:
    if j["total"] == maior_total:
        print(f"- {j['nome']}: {j['total']} pontos")

print("\n Mais regulares (menor diferença):")
for j in jogadores:
    if j["diferenca"] == menor_diff:
        print(f"- {j['nome']}: diferença = {j['diferenca']}")
print(" Exercício 9 concluído!\n")



 EXERCÍCIO 10 — Agenda de Contatos Sem Duplicidade

print("="*60)
print("EXERCÍCIO 10 — Agenda de Contatos")
print("="*60)

agenda = []

while True:
    nome = input("Nome (digite FIM para encerrar): ").strip()
    if nome.upper() == "FIM":
        break

     Verificar duplicidade
    existe = False
    for c in agenda:
        if c["nome"] == nome:
            existe = True
            break

    if existe:
        print(f"  Contato '{nome}' já existe! Não permitido duplicar.")
        continue

    telefone = input("Telefone: ")
    agenda.append({"nome": nome, "telefone": telefone})
    print(f" Contato '{nome}' cadastrado!")

 Consulta
print("\n--- Consulta de Contatos ---")
while True:
    busca = input("Digite nome para buscar (ou FIM para sair): ").strip()
    if busca.upper() == "FIM":
        break

    encontrado = False
    for c in agenda:
        if c["nome"] == busca:
            print(f" {c['nome']}: {c['telefone']}")
            encontrado = True
            break
    if not encontrado:
        print(f" Contato '{busca}' não encontrado.")

print(" Exercício 10 concluído!\n")



 EXERCÍCIO 11 — Pesquisa de Votos

print("="*60)
print("EXERCÍCIO 11 — Pesquisa de Votos")
print("="*60)

votos = {}
total_votos = 0

while True:
    opcao = input("Voto (1 a 4, 0 para encerrar): ").strip()
    if opcao == "0":
        break
    if opcao not in ["1", "2", "3", "4"]:
        print("  Código inválido! Ignorado.")
        continue

    if opcao in votos:
        votos[opcao] += 1
    else:
        votos[opcao] = 1
    total_votos += 1

 Garantir que todas opções existem
for cod in ["1", "2", "3", "4"]:
    if cod not in votos:
        votos[cod] = 0

 Encontrar maior votação sem max()
maior = votos["1"]
for cod in votos:
    if votos[cod] > maior:
        maior = votos[cod]

vencedores = [cod for cod in votos if votos[cod] == maior]

print("\n Resultado da Votação:")
for cod, qtd in votos.items():
    perc = (qtd / total_votos * 100) if total_votos > 0 else 0
    print(f"Opção {cod}: {qtd} votos ({perc:.1f}%)")

print(f"\nTotal de votos válidos: {total_votos}")
print(f" Opção(ões) vencedora(s): {', '.join(vencedores)} com {maior} votos")
print(" Exercício 11 concluído!\n")



 EXERCÍCIO 12 — Histórico de Movimentações

print("="*60)
print("EXERCÍCIO 12 — Histórico de Movimentações")
print("="*60)

estoque = {}
historico = []

 Cadastrar produto inicial
cod_prod = input("Código do produto: ")
nome_prod = input("Nome: ")
preco_prod = float(input("Preço: "))
qtd_inicial = int(input("Quantidade inicial: "))
estoque[cod_prod] = {
    "nome": nome_prod, "preco": preco_prod, "quantidade": qtd_inicial
}

 8 movimentações
for mov in range(8):
    print(f"\n--- Movimentação {mov+1} ---")
    cod = input("Código: ")
    tipo = input("Entrada (E) ou Saída (S): ").upper()
    qtd = int(input("Quantidade: "))

    if cod not in estoque:
        print(" Produto não cadastrado! Movimentação recusada.")
        historico.append((cod, tipo, qtd))
        continue

    if tipo == "E":
        estoque[cod]["quantidade"] += qtd
        historico.append((cod, tipo, qtd))
        print(f" Entrada realizada. Estoque: {estoque[cod]['quantidade']}")
    elif tipo == "S":
        if estoque[cod]["quantidade"] >= qtd:
            estoque[cod]["quantidade"] -= qtd
            historico.append((cod, tipo, qtd))
            print(f" Saída realizada. Estoque: {estoque[cod]['quantidade']}")
        else:
            print(" Estoque insuficiente! Saída recusada.")
            historico.append((cod, "RECUSADA", qtd))
    else:
        print(" Tipo inválido!")
        historico.append((cod, "INVÁLIDA", qtd))

 Relatório final
print("\n Histórico de Movimentações:")
for i, m in enumerate(historico, 1):
    print(f"{i}. Cód: {m[0]} | Tipo: {m[1]} | Qtd: {m[2]}")

print("\n Estoque Final:")
for cod, p in estoque.items():
    print(f"{cod} — {p['nome']}: {p['quantidade']} unidades")
print(" Exercício 12 concluído!\n")



EXERCÍCIO 13 — Análise de Vendas por Vendedor

print("="*60)
print("EXERCÍCIO 13 — Análise de Vendas por Vendedor")
print("="*60)

vendedores = []

for _ in range(5):
    nome = input("Nome do vendedor: ")
    vendas = []
    for i in range(4):
        vendas.append(float(input(f"Venda {i+1}: ")))

    Total sem usar sum()
    total = 0
    for v in vendas:
        total += v

    vendedores.append({"nome": nome, "vendas": vendas, "total": total})

Total geral
total_geral = 0
for v in vendedores:
    total_geral += v["total"]
media_equipe = total_geral / len(vendedores)

Maior total sem usar max()
maior_total = vendedores[0]["total"]
for v in vendedores:
    if v["total"] > maior_total:
        maior_total = v["total"]

print("\n Vendas por Vendedor:")
for v in vendedores:
    print(f"{v['nome']} — Vendas: {v['vendas']} | Total: R$ {v['total']:.2f}")

print(f"\n Total geral de vendas: R$ {total_geral:.2f}")
print(f" Média da equipe: R$ {media_equipe:.2f}")

print("\n Vendedor(es) com maior total:")
for v in vendedores:
    if v["total"] == maior_total:
        print(f"- {v['nome']}: R$ {v['total']:.2f}")

print("\n Vendedor(es) acima da média da equipe:")
for v in vendedores:
    if v["total"] >= media_equipe:
        print(f"- {v['nome']}: R$ {v['total']:.2f}")
print(" Exercício 13 concluído!\n")



 EXERCÍCIO 14 — Catálogo e Pedidos

print("="*60)
print("EXERCÍCIO 14 — Catálogo e Pedidos")
print("="*60)

catalogo = []

 Cadastrar 5 produtos
for _ in range(5):
    cod = input("Código: ")
    nome = input("Nome: ")
    preco = float(input("Preço: "))
    qtd_est = int(input("Estoque: "))
    catalogo.append({
        "codigo": cod, "nome": nome, "preco": preco, "estoque": qtd_est
    })

 Cadastrar 6 pedidos antes de processar
pedidos = []
for p in range(6):
    cod_ped = input(f"\nPedido {p+1} — Código: ")
    qtd_ped = int(input(f"Quantidade: "))
    pedidos.append((cod_ped, qtd_ped))

 Processar pedidos
print("\n--- Processamento de Pedidos ---")
valor_total_atendidos = 0
valor_total_recusados = 0

for cod, qtd in pedidos:
    prod = None
    for p in catalogo:
        if p["codigo"] == cod:
            prod = p
            break

    if not prod:
        print(f" Pedido {cod}: Produto não encontrado — RECUSADO")
        valor_total_recusados += 0
        continue

    if prod["estoque"] >= qtd:
        prod["estoque"] -= qtd
        valor = prod["preco"] * qtd
        valor_total_atendidos += valor
        print(f" Pedido {prod['nome']} x{qtd} — ATENDIDO | R$ {valor:.2f}")
    else:
        print(f" Pedido {prod['nome']} x{qtd} — ESTOQUE INSUFICIENTE — RECUSADO")
        valor_total_recusados += 0

print(f"\n Valor total de pedidos atendidos: R$ {valor_total_atendidos:.2f}")
print(f" Estoque Final:")
for p in catalogo:
    print(f"{p['codigo']} — {p['nome']}: {p['estoque']} unidades")
print(" Exercício 14 concluído!\n")



 EXERCÍCIO 15 — Sistema de Vendas e Relatório Final

print("="*60)
print("EXERCÍCIO 15 — Sistema de Vendas e Relatório Final")
print("="*60)

produtos = []
historico_vendas = []

 Cadastrar 5 produtos
for _ in range(5):
    cod = input("Código: ")
    nome = input("Nome: ")
    preco = float(input("Preço: "))
    qtd_est = int(input("Quantidade em estoque: "))
    produtos.append({
        "codigo": cod, "nome": nome, "preco": preco, "estoque": qtd_est
    })

 Processar vendas até código 0
unidades_vendidas = 0
valor_total_vendas = 0

while True:
    cod_venda = input("\nCódigo para vender (0 para encerrar): ").strip()
    if cod_venda == "0":
        break

    qtd_venda = int(input("Quantidade: "))

     Localizar produto
    prod = None
    for p in produtos:
        if p["codigo"] == cod_venda:
            prod = p
            break

    if not prod:
        print(" Produto não existe! Venda cancelada.")
        continue

    if prod["estoque"] >= qtd_venda:
        prod["estoque"] -= qtd_venda
        valor_venda = prod["preco"] * qtd_venda
        unidades_vendidas += qtd_venda
        valor_total_vendas += valor_venda
        historico_vendas.append((cod_venda, qtd_venda, valor_venda))
        print(f" Venda realizada! {prod['nome']} x{qtd_venda} — R$ {valor_venda:.2f}")
    else:
        print(" Estoque insuficiente! Venda cancelada.")

Encontrar produto de maior valor sem usar max()
maior_valor = produtos[0]["preco"]
prod_maior_valor = produtos[0]
for p in produtos:
    if p["preco"] > maior_valor:
        maior_valor = p["preco"]
        prod_maior_valor = p

 Relatório final
print("\n" + "="*60)
print(" RELATÓRIO FINAL")
print("="*60)

print("\n Estoque Restante:")
for p in produtos:
    print(f"{p['codigo']} — {p['nome']}: {p['estoque']} unidades")

print("\n Histórico de Vendas (tupla):")
for cod, qtd, val in historico_vendas:
    print(f"Cód: {cod} | Qtd: {qtd} | Valor: R$ {val:.2f}")

tupla_resumo = (unidades_vendidas, valor_total_vendas)
print(f"\n Resumo (tupla):")
print(f"Unidades vendidas: {tupla_resumo[0]}")
print(f"Valor total das vendas: R$ {tupla_resumo[1]:.2f}")

print(f"\n Produto de maior valor unitário: {prod_maior_valor['nome']} — R$ {prod_maior_valor['preco']:.2f}")
print("\n Exercício 15 concluído!")
print("\n TODOS OS EXERCÍCIOS CONCLUÍDOS COM SUCESSO! ")
print("="*60)
```

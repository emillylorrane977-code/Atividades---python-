```python


 1. Ler 6 números e mostrar o maior e o menor
print("\n=== Exercício 1 ===")
primeiro = int(input("Digite um número: "))
maior = primeiro
menor = primeiro

for i in range(5):
    numero = int(input("Digite um número: "))
    if numero > maior:
        maior = numero
    if numero < menor:
        menor = numero

print(f"Maior número: {maior}")
print(f"Menor número: {menor}")


 2. Ler 5 notas e mostrar média, aprovados e reprovados
print("\n=== Exercício 2 ===")
soma = 0
aprovados = 0
reprovados = 0

for i in range(5):
    nota = float(input(f"Nota {i+1}: "))
    soma += nota
    if nota >= 7:
        aprovados += 1
    else:
        reprovados += 1

media = soma / 5
print(f"Média da turma: {media:.1f}")
print(f"Notas ≥ 7: {aprovados} aluno(s)")
print(f"Notas < 7: {reprovados} aluno(s)")


3. Validando uma nota entre 0 e 10
print("\n=== Exercício 3 ===")
nota = float(input("Digite uma nota entre 0 e 10: "))
while nota < 0 or nota > 10:
    print("Nota inválida! Tente novamente.")
    nota = float(input("Digite uma nota entre 0 e 10: "))
print("Nota válida registrada:", nota)


4. Cálculo de valor de estaciona
print("\n=== Exercício 4 ===")
horas = int(input("Tempo em horas: "))
if horas <= 2:
    valor = 8.00
elif horas <= 5:
    valor = 8.00 + (horas - 2) * 3.00
else:
    valor = 17.00 + (horas - 5) * 2.00
print(f"Valor a pagar: R$ {valor:.2f}")


5. Máquina de venda de ingresso (R$ 25,00)

print("\n=== Exercício 5 ===")
total = 0.0
PRECO_INGRESSO = 25.00
while total < PRECO_INGRESSO:
    valor = float(input("Insira valor: R$ "))
    total += valor
troco = total - PRECO_INGRESSO
print(f"Total pago: R$ {total:.2f}")
print(f"Troco: R$ {troco:.2f}")


6. Maior valor e sua posição

print("\n=== Exercício 6 ===")
maior = None
posicao_maior = 0
for posicao in range(1, 9):
    num = int(input(f"Número {posicao}: "))
    if maior is None or num > maior:
        maior = num
        posicao_maior = posicao
print(f"Maior valor: {maior} — Posição: {posicao_maior}")


7. Segundo maior valor

print("\n=== Exercício 7 ===")
maior = None
segundo_maior = None
for i in range(1, 9):
    num = int(input(f"Número {i}: "))
    if maior is None:
        maior = num
    elif segundo_maior is None:
        if num > maior:
            segundo_maior = maior
            maior = num
        else:
            segundo_maior = num
    else:
        if num > maior:
            segundo_maior = maior
            maior = num
        elif num > segundo_maior:
            segundo_maior = num
print(f"Maior: {maior} — Segundo maior: {segundo_maior}")


8. Detecta valores consecutivos iguais

print("\n=== Exercício 8 ===")
tem_igual = False
anterior = int(input("Número 1: "))
for i in range(2, 9):
    atual = int(input(f"Número {i}: "))
    if atual == anterior:
        tem_igual = True
    anterior = atual
if tem_igual:
    print("Existem valores consecutivos iguais.")
else:
    print("Não existem valores consecutivos iguais.")


9. Controle de saldo bancário

print("\n=== Exercício 9 ===")
saldo = 500.00
for i in range(1, 7):
    print(f"\nOperação {i}")
    tipo = int(input("1 - Depósito | 2 - Saque: "))
    valor = float(input("Valor: R$ "))
    if tipo == 1:
        saldo += valor
        print("Depósito realizado.")
    elif tipo == 2:
        if valor <= saldo:
            saldo -= valor
            print("Saque realizado.")
        else:
            print("Saldo insuficiente — saque não realizado.")
    else:
        print("Operação inválida.")
print(f"\nSaldo final: R$ {saldo:.2f}")


10. Cálculo de cédulas para saque

print("\n=== Exercício 10 ===")
valor = int(input("Valor do saque: R$ "))
cedulas = [100, 50, 20, 10, 5, 2, 1]
print("Quantidade de cédulas:")
for c in cedulas:
    qtd = valor // c
    valor = valor % c
    if qtd > 0:
        print(f"R$ {c:3} → {qtd}")


11. Verifica se número é primo

print("\n=== Exercício 11 ===")
n = int(input("Digite um número inteiro maior que 1: "))
e_primo = True
for d in range(2, int(n**0.5) + 1):
    if n % d == 0:
        e_primo = False
        break
if e_primo:
    print(f"{n} é um número primo.")
else:
    print(f"{n} não é um número primo.")


12. Mostra todos os divisores de um número

print("\n=== Exercício 12 ===")
n = int(input("Digite um número inteiro positivo: "))
divisores = []
for d in range(1, n + 1):
    if n % d == 0:
        divisores.append(d)
print(f"Divisores de {n}: {divisores}")
print(f"Quantidade de divisores: {len(divisores)}")


13. Lista números primos em um intervalo

print("\n=== Exercício 13 ===")
inicio = int(input("Valor inicial: "))
fim = int(input("Valor final: "))
primos = []
for num in range(inicio, fim + 1):
    if num < 2:
        continue
    primo = True
    for d in range(2, int(num**0.5) + 1):
        if num % d == 0:
            primo = False
            break
    if primo:
        primos.append(num)
print(f"Números primos: {primos}")
print(f"Quantidade encontrada: {len(primos)}")


14. Jogo — adivinhe o número secreto

print("\n=== Exercício 14 ===")
SECRETO = 37
MAX_TENTATIVAS = 5
acertou = False
for tentativa in range(1, MAX_TENTATIVAS + 1):
    palpite = int(input(f"Tentativa {tentativa} de {MAX_TENTATIVAS}: "))
    if palpite == SECRETO:
        print("Parabéns! Você acertou!")
        acertou = True
        break
    elif palpite > SECRETO:
        print("O número secreto é menor.")
    else:
        print("O número secreto é maior.")
if not acertou:
    print(f"Fim das tentativas. O número era {SECRETO}.")


15. Caixa de atendimento completo

print("\n=== Exercício 15 ===")
saldo = 200.00
qtd_depositos = 0
qtd_saques = 0
qtd_recusados = 0
total_entrada = 0.00
total_saida = 0.00

while True:
    print("\nMENU:")
    print("1 — Depósito")
    print("2 — Saque")
    print("0 — Encerrar")
    opcao = int(input("Escolha a operação: "))
    if opcao == 0:
        break
    if opcao == 1:
        valor = float(input("Valor do depósito: R$ "))
        saldo += valor
        qtd_depositos += 1
        total_entrada += valor
    elif opcao == 2:
        valor = float(input("Valor do saque: R$ "))
        if valor <= saldo:
            saldo -= valor
            qtd_saques += 1
            total_saida += valor
        else:
            print("Saldo insuficiente — saque recusado.")
            qtd_recusados += 1
    else:
        print("Opção inválida!")

print("\n" + "="*40)
print(f"Saldo final:         R$ {saldo:.2f}")
print(f"Depósitos:           {qtd_depositos}")
print(f"Saques realizados:   {qtd_saques}")
print(f"Saques recusados:    {qtd_recusados}")
print(f"Total de entradas:   R$ {total_entrada:.2f}")
print(f"Total de saídas:     R$ {total_saida:.2f}")
print("="*40)
print("\n=== Fim da execução ===")
```
